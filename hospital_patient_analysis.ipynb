```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("C:\\Users\\acer\\PYTHON\\Datasets\\smart_hospital_appointment_dataset.csv")
#print(df)
```

```python
print(df.columns)
```

```python
print("Duplicated rows:")
print(df.duplicated().sum())
print("Missing values:")
print(df.isna().sum())
```

```python
df= df.dropna()
print(df)
print(df.isna().sum())
```

```python
#PATIENT ANALYSIS
print("Total Patients:", df.shape[0])
print("\nMale vs Female Patients")
print(df["gender"].value_counts())
print("\nAge Distribution")
print(df["age"].describe())
print("Average Age:", df["age"].mean())
print("Youngest Patient:", df["age"].min())
print("Oldest Patient:", df["age"].max())
```

```python
# DEPARTMENT ANALYSIS
print("\nPatients per Department")
print(df["department"].value_counts())
print("Most Visited Department:", df["department"].mode()[0])
print("Least Visited Department:", df["department"].value_counts().idxmin())
```

```python
# CITY ANALYSIS
print("\nPatients from Each City")
print(df["city"].value_counts())
print("Most Common City:", df["city"].mode()[0])
print("Least Common City:", df["city"].value_counts().idxmin())
```

```python
# WAITING TIME ANALYSIS
print("\nAverage Waiting Time:", df["waiting_time_minutes"].mean())
print("Minimum Waiting Time:", df["waiting_time_minutes"].min())
print("Maximum Waiting Time:", df["waiting_time_minutes"].max())
```

```python
# BILLING ANALYSIS
print("\nAverage Consultation Fee:", df["consultation_fee"].mean())
print("Average Medicine Cost:", df["medicine_cost"].mean())
print("Average Test Cost:", df["test_cost"].mean())
print("Average Total Bill:", df["total_bill"].mean())
print("Highest Bill:", df["total_bill"].max())
print("Lowest Bill:", df["total_bill"].min())
```

```python
#1.	Patient Gender Distribution (Bar Chart) 
gender = df["gender"].value_counts()
plt.figure(figsize=(6,5))
plt.bar(gender.index, gender.values, color=["blue", "pink"])
plt.title("Patient Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Patient Age Distribution (Histogram) 
plt.figure(figsize=(8,5))
plt.hist(df["age"],
         bins=10,
         color="orange",
         edgecolor="black")
plt.title("Patient Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Department-wise Patients (Bar Chart) 
department = df["department"].value_counts()
plt.figure(figsize=(8,5))
plt.bar(department.index,
        department.values,
        color="green")
plt.title("Patients per Department")
plt.xlabel("Department")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
#City-wise Patients (Bar Chart) 
city = df["city"].value_counts()
plt.figure(figsize=(10,5))
plt.bar(city.index,
        city.values,
        color="violet")
plt.title("Patients from Each City")
plt.xlabel("City")
plt.ylabel("Number of Patients")
plt.xticks(rotation=45)
plt.show()
```

```python
#Waiting Time Distribution (Histogram) 
plt.figure(figsize=(8,5))
plt.hist(df["waiting_time_minutes"],
         bins=10,
         color="skyblue",
         edgecolor="black",
         linewidth=1)
plt.title("Waiting Time Distribution")
plt.xlabel("Waiting Time (Minutes)")
plt.ylabel("Number of Patients")
plt.show()
```

```python
#Total Bill Distribution (Histogram) 
plt.figure(figsize=(8,5))
plt.hist(df["total_bill"],
         bins=10,
         color="red",
         edgecolor="black",
         linewidth=1,
         alpha=0.8)
plt.title("Total Bill Distribution")
plt.xlabel("Total Bill Amount")
plt.ylabel("Number of Patients")
plt.show()
```
